````python
import cv2
import mss
import numpy as np

capture = mss.MSS()
monitors = capture.monitors[1:]
left_monitor = min(monitors, key=lambda monitor: (monitor['left'], monitor['top']))
print('Monitors:', monitors)
print('Left monitor:', left_monitor)
````

````python
# Capture one frame only; this cell does not press any key.
frame = np.array(capture.grab(left_monitor))[:, :, :3]
print('Captured BGR frame:', frame.shape)
assert frame.shape[1] == left_monitor['width']
assert frame.shape[0] == left_monitor['height']
````
